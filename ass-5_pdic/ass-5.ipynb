```python
# 4) Write a Python program to calculate the total, highest value, and lowest value from a tuple 
# of integers without using the built-in sum(), max(), and min() functions
t = (10, 5, 20, 8, 15)

total = 0
highest = t[0]
lowest = t[0]

for i in t:
    total += i

    if i > highest:
        highest = i

    if i < lowest:
        lowest = i

print("Total =", total)
print("Highest =", highest)
print("Lowest =", lowest)
```

```python
#5) Write a Python program to filter a tuple . n = (3, 14, 7, 22, 9, 41, 18, 5),
#  keep only values greater than 10 
n = (3, 14, 7, 22, 9, 41, 18, 5)

result = ()

for i in n:
    if i > 10:
        result += (i,)

print(result)
```

```python
#6) Write a Python program to determine how many elements are in a set without 
# using the built-in len() function. s = {"cat", "dog", "bird", "fish"}
s = {"cat", "dog", "bird", "fish"}

count = 0

for i in s:
    count += 1

print("Count =", count)
```
